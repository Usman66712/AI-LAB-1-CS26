Program 1: Even or Odd

Objective: To check whether a number is even or odd using an if-else statement.



```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

Program 2: Positive or Negative

Objective: To determine whether a number is positive, negative, or zero.



```python
num = int(input("Enter a number: "))

if num > 0:
    print("The number is positive")
elif num < 0:
    print("The number is negative")
else:
    print("The number is zero")
```

### Program 3: Greater Number
**Objective:** To compare two numbers and find the greater one.



```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

if a > b:
    print("First number is greater")
else:
    print("Second number is greater or both are equal")
```

Objective: To check whether a person is eligible to vote based on age

```python
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")
```

Program 5: Pass or Fail

Objective: To check whether a student has passed or failed based on marks.

```python
marks = int(input("Enter your marks: "))

if marks >= 50:
    print("You have passed")
else:
    print("You have failed")
```
