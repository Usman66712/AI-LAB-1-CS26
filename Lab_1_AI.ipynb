Program 1: Even or Odd

Objective: To check whether a number is even or odd using an if-else statement.



```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

Program 2: Positive or Negative

Objective: To determine whether a number is positive, negative, or zero.



```python
num = int(input("Enter a number: "))

if num > 0:
    print("The number is positive")
elif num < 0:
    print("The number is negative")
else:
    print("The number is zero")
```

### Program 3: Greater Number
**Objective:** To compare two numbers and find the greater one.



```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

if a > b:
    print("First number is greater")
else:
    print("Second number is greater or both are equal")
```

Objective: To check whether a person is eligible to vote based on age

```python
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")
```

Program 5: Pass or Fail

Objective: To check whether a student has passed or failed based on marks.

```python
marks = int(input("Enter your marks: "))

if marks >= 50:
    print("You have passed")
else:
    print("You have failed")
```

Python basics — all in one program

```python

# Student Marks Program

name = input("Enter your name: ")
age = int(input("Enter your age: "))

print("Welcome", name)
print("Your age is", age)

subjects = ["Python", "Math", "English"]
marks = []

for subject in subjects:
    m = int(input("Enter marks in " + subject + ": "))
    marks.append(m)

total = marks[0] + marks[1] + marks[2]
average = total / 3

print("Total marks:", total)
print("Average marks:", average)

if average >= 80:
    print("Grade A")
elif average >= 70:
    print("Grade B")
elif average >= 60:
    print("Grade C")
elif average >= 50:
    print("Grade D")
else:
    print("Fail")

# checking if student passed
if average >= 50 and age >= 18:
    print("You passed and you are an adult")
else:
    print("Keep working hard")

# function
def show_name(n):
    print("Student name is", n)

show_name(name)

# tuple
info = (name, age)
print("Student information:", info)

# dictionary
student = {
    "name": name,
    "age": age,
    "marks": marks
}

print(student)

# string operations
print(name.upper())
print(name.lower())
print("Name length:", len(name))

# while loop
i = 1
while i <= 3:
    print("Practice makes perfect")
    i = i + 1

# list comprehension
passed = [m for m in marks if m >= 50]
print("Passing marks:", passed)

# exception handling
try:
    num = int(input("Enter a number: "))
    print("Double of number:", num * 2)
except ValueError:
    print("Please enter a number")

print("Program ended")

```
